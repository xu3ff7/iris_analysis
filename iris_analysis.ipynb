يتم إنشاء مجلد output لحفظ ملفات نتائج التحليل.

```python
species_summary = df.groupby("species").mean(numeric_only=True)

species_summary.to_csv("output/species_summary.csv")
```

يتم حفظ الزهور التي تحقق شرط sepal_length > 6 في ملف CSV جديد.
